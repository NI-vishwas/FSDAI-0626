```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
spdf.groupby('school_type')['overall_score'].mean()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='school_type', y='overall_score', data=spdf)
plt.title('Overall Score by School Type')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='school_type', y='study_hours', data=spdf)
plt.title('Average Study Hours by School Type')
plt.show()
```
