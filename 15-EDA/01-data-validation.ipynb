```python
import pydantic
import sqlalchemy as sa
import pymysql
```

```python
from sqlalchemy import create_engine, Column, Integer, String, Float, DateTime, Boolean
from sqlalchemy.orm import declarative_base, sessionmaker
from dotenv import load_dotenv
import os

load_dotenv()

DATABASE_USER = os.getenv("DATABASE_USER")
DATABASE_PASSWORD = os.getenv("DATABASE_PASSWORD")
DATABASE_HOST = os.getenv("DATABASE_HOST")
DATABASE_NAME = os.getenv("DATABASE_NAME")

DATABASE_URL = f"mysql+pymysql://{DATABASE_USER}:{DATABASE_PASSWORD}@{DATABASE_HOST}/{DATABASE_NAME}"

engine = create_engine(DATABASE_URL)
SessionLocal = sessionmaker(autocommit=False, autoflush=False, bind=engine)
Base = declarative_base()
```

```python
class Employee(Base):
    __tablename__ = "Employees"

    EmployeeID = Column(Integer, primary_key=True, index=True)
    LastName = Column(String(20), nullable=False)
    FirstName = Column(String(10), nullable=False)
    BirthDate = Column(DateTime)
    Notes = Column(String)
```

```python
def get_db():
    db = SessionLocal()
    try:
        yield db
    finally:
        db.close()
```

```python
def get_employee_and_validate(db, employee_id: int):
    employee = db.query(Employee).filter(Employee.EmployeeID == employee_id).first()
    if not employee:
        raise ValueError(f"Employee with ID {employee_id} not found.")
    return employee
```

```python
emp = get_employee_and_validate(next(get_db()), 1)
print(emp.FirstName, emp.LastName, emp.BirthDate, emp.Notes)
```
