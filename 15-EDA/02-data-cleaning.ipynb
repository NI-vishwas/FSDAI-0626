```python
import pandas as pd
import sqlalchemy as sa
import pymysql
```

```python
d1 = {
    "Name": ["Alice", None, "", "David"],
    "Age": [25, 30, 35, "40"],
    "City": ["New York", None, "Chicago", None]
}
```

```python
df = pd.DataFrame(d1)
df.head()
```

```python
df.info()
```

How to Handle Missing data                    

1. Drop those - Data Loss
2. Fill with default values

```python
df1 = df.copy()
df1.dropna(inplace=True)
df1
```

```python
df2 = df.copy()
df2.fillna({"Name": "Unknown", "City": "Unknown", "Age": 0}, inplace=True)
df2
```

```python
df2["Age"] = df2["Age"].astype(int)
df2.info()
```
